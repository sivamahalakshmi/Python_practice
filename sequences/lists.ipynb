```python
#1.Sum of all elements: Write a Python function called sum_list that takes a list of numbers as an argument and returns the sum of all the items in the list.
n=input().split()
sum=0
for i in n:
    sum+= int(i)
print(sum)
```

```python
#8.Count occurrences: Write a Python function that takes a list and an element as input, and returns the number of times that element appears in the list.
l=input().split()
key=input()
count=0
for i in l:
    if i in key:
        count+=1
print(count)
```

```python
#14.Print the largest and smallest number in a list.
l=list(map(int,input().split()))
small=l[0]
large=l[0]
for i in l:
    if i<small:
        small=i
    if i>large:
        large=i
print("Smallest:",small)
print("Largest:",large)
```

```python
#15.Add a number to a list only if it is not already present. Hint: Use if number not in list:
l=list(map(int,input().split()))
n=int(input())
if n not in l:
    l.append(n)
print(l)
```

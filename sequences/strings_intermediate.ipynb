```python
#1. Count Frequency of Each Character in a String
# Input: Enter a string: programming
# Output: p:1 r:2 o:1 g:2 a:1 m:2 i:1 n:1
s=input()
f={}
for ch in s:
    if ch in f:
        f[ch]+=1
    else:
        f[ch]=1
for keys in f:
    print(f"{keys}:{f[keys]}", end=" ")
```

```python
# 2. Remove Duplicate Characters from a String
# Input: Enter a string: banana
# Output: String after removing duplicates: ban
# Hint: Check if character already exists
s=input()
final=""
for i in s:
    if i not in final:
        final+=i
print(final)
```

```python
#3. Find the Longest Word in a Sentence
# Input: Enter a sentence: Java programming is very powerful
# Output: Longest word: programming
# Hint: Use split() and length comparison
s=input().split()
long_no=0
long_word=""
for i in s:
    if len(i)>long_no:
        long_no=len(i)
        long_word=i
    else:
        long_word=long_word
print(long_word)
```

```python
# 4. Count Occurrence of a Word in a Sentence
# Input: Enter a sentence: Java is easy and Java is powerful
# Enter word: Java
# Output: Word occurrence: 2
# Hint: Split sentence and compare
s=input().split()
word=input()
c=0
for i in s:
    if i==word:
        c+=1
print(c)

```

```python
# 5. Check if Two Strings are Anagrams
# Input: Enter first string: listen Enter second string: silent
# Output: Strings are anagrams
# Hint: Sort character arrays
s1=input().split()
s2=input().split()
l1=s1.sort()
l2=s2.sort()
if l1==l2:
    print("anagram")
else:
    print("not anagram")
```

```python
# 6. Toggle Case of Each Character
# Input: Enter a string: JaVa
# Output: Toggled string: jAvA
# Hint: Use Character class methods
s=input()
print(s.swapcase())
```

```python
# 8. Compress a String
# Input: Enter a string: aaabbc
# Output: a3b2c1
# Hint: Count consecutive characters
s=input()
res=""
for i in s:
    if i not in res:
        res+=i+str(s.count(i))
print(res)
```

```python
#9. Reverse Words in a Sentence
#Input: Enter a sentence: Java is easy
#Output: avaJ si ysae
s=input().split()
for i in s:
    print(i[::-1],end=" ")
```

```python
#10. Check Valid Email Format
#Input: Enter email: student@gmail.com
#Output: Valid email

s=input()
if "@" in s and "." in s:
    print("Valid email")
else:
    print("Invalid email")
```

```python
#11. Find All Substrings of a String
#Input: Enter a string: abc
#Output: a ab abc b bc c
s=input()
for i in range(len(s)):
    for j in range(i+1,len(s)+1):
        print(s[i:j],end=" ")
```

```python
#12. Capitalize First Letter of Each Word
#Input: Enter a sentence: java is powerful
#Output: Java Is Powerful
s=input().split()
for i in s:
    print(i.capitalize(),end=" ")
```

```python
#13. Remove Special Characters from a String
#Input: Enter a string: Ja@va#123!
#Output: Java123

s=input()
res=""
for i in s:
    if i.isalnum():
        res+=i
print(res)
```

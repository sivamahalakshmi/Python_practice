```python
#3.Count Word Frequency
#Ask the user to enter a short sentence. Count how many times each word appears using a dictionary.
s=input()
freq={}
a=s.split()
for i in a:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
print(freq)

```

```python
#3.Student Marks Dictionary
#Create a dictionary of student names and their marks.
#Print the student with the highest marks.
#Print the student with the lowest marks.
#Find the average marks of the class.
marks={'a':90,'b':88,'c':40,'d':44,'e':99,'f':60}
print(max(marks))
print(min(marks))
avg=0
for i in marks.values():
    avg+=i
avg=avg/len(marks)
print(avg)
```

```python
# 4. Nested Dictionary – Employee Records
# Create a dictionary of 3 employees where each employee has:
# {
#   'emp_id': 101,
#   'name': 'Alice',
#   'department': 'HR',
#   'salary': 50000
# }
# Store them inside another dictionary with keys "emp1", "emp2", "emp3".
# Print the name of the employee who has the highest salary.
# Add a new employee "emp4".

employees={
    "emp1":{'emp_id':101,'name':'alice','department':'HR','salary':50000},
    "emp2":{'emp_id':102,'name':'bob','department':'IT','salary':60000},
    "emp3":{'emp_id':103,'name':'charlie','department':'Finance','salary': 55000}
}

max_salary=0
max_name=""
for i in employees.values():
    if i['salary']>max_salary:
        max_salary=i['salary']
        max_name=i['name']
print(max_salary)
print(max_name)

employees["emp4"]={'emp_id':104,'name':'david','department':'management','salary': 30000}
print(employees)
```

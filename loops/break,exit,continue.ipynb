```python
import sys
def process_age(age):
   if age < 0:
       raise ValueError("Age cannot be negative")
   elif age < 18:
       sys.exit("Age less than 18. Exiting program.")
   else:
       print("Valid age:", age)
try:
   process_age(10) # Will exit here
   print("This won't run if sys.exit() is called")
except ValueError as e:
   print("Error:", e)
   sys.exit(1) # Exit with non-zero code for error
```

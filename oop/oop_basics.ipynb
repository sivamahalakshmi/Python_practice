```python
# 1. Student Class
# Create a Student class with:
# •	Attributes: name, roll_number, marks
# •	Method to display student details
# •	Method to calculate grade based on marks
# (90+ = A, 75–89 = B, 50–74 = C, below 50 = Fail)
class Student:
    def __init__(self,name,r_no,marks):
        self.name=name
        self.roll_number=r_no
        self.marks=marks
    def details(self):
        print(self.name,self.roll_number,self.marks)
    def grade(self):
        if self.marks>=90:
            print("A")
        elif self.marks>75:
            print("B")
        elif self.marks>50:
            print("C")
        else:
            print("D")
s1=Student("maha",53,99)
s2=Student("nivi",32,77)
s3=Student("sai",43,50)

s1.details()
s1.grade()

s2.details()
s2.grade()

s3.details()
s3.grade()
```

```python
# 2. Rectangle Class
# Create a Rectangle class with:
# •	Attributes: length, width
# •	Method to calculate area
# •	Method to calculate perimeter
# •	Method to check if it is a square

class RectangleClass:
    def __init__(self,len,wid):
        self.length=len
        self.width=wid
    def area(self):
        return (self.length*self.width)
    def perimeter(self):
        return 2*(self.length+self.width)
    def square(self):
        if self.length==self.width:
            print("square")
        else:
            print("rectangle")
x=RectangleClass(2,3)
print(x.area())
print(x.perimeter())
x.square()
```

```python
# 4. Car Class
# Create a Car class with:
# •	Attributes: brand, model, year, speed
# •	Method to accelerate (increase speed)
# •	Method to brake (decrease speed without going below 0)
# •	Method to display car details
class Car:
    def __init__(self,brand,model,year,speed):
        self.brand=brand
        self.model=model
        self.year=year
        self.speed=speed
    def accelerate(self):
        acc=int(input("enter acceleration speed"))
        self.speed+=acc
        return self.speed

    def decelerate(self):
        dec=int(input("enter deceleration speed"))
        self.speed-=dec
        return self.speed

    def details(self):
        print(self.brand,self.model,self.year,self.speed)
c1=Car('Mahindra','Thar','2026',120)
print(c1.accelerate())
print(c1.decelerate())
c1.details()
```

```python
# 5. Employee Class
# Create an Employee class with:
# •	Attributes: name, salary
# •	Method to calculate annual salary
# •	Method to apply a bonus (percentage-based)
# •	Method to display employee information
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def annual(self):
        annual=12*self.salary
        print(annual)
    def bonus(self):
        percentage=int(input("enter bonus percentage"))
        bonus=(percentage/100)*self.salary
        self.salary+=bonus
        print(self.salary)
    def info(self):
        print(self.name, self.salary)
e1=Employee("maha",50000)
e1.annual()
e1.bonus()
e1.info()
```

```python
# 6. Book Class
# Create a Book class with:
# •	Attributes: title, author, price
# •	Method to apply discount (percentage)
# •	Method to display book details
# •	Create multiple book objects and display their details
class Book:
    def __init__(self,title,author,price):
        self.title=title
        self.author=author
        self.price=price
    def discount(self):
        percentage=int(input("enter discount percentage"))
        discount=(percentage/100)*self.price
        self.price-=discount
        print(self.price)
    def details(self):
        print(self.title,self.author,self.price)
b1=Book("Ponniyin selvan", "Kalki",300)
b2=Book("Sivakamiyin sabatham", "Kalki",400)
b1.discount()
b1.details()
b2.details()
```

```python
# 7. Circle Class
# Create a Circle class with:
# •	Attribute: radius
# •	Method to calculate area
# •	Method to calculate circumference
# •	Method to compare two circles and print which one is bigger
import math
class Circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        area=math.pi*(self.radius**2)
        return area
    def perimeter(self):
        perimeter=2*math.pi*self.radius
        print(perimeter)
    def bigger(self,t):
        if t.area()>self.area():
            print("c2 is bigger")
        else:
            print("c1 is bigger")
c1=Circle(5)
c2=Circle(10)
print(c1.area())
c1.perimeter()
c1.bigger(c2)
```

```python
#8. ShoppingCart Class
#Attribute: items (dictionary: item_name→price)
#Method to add item
#Method to remove item
#Method to calculate total price
#Method to display cart items

class ShoppingCart:
    def __init__(self):
        self.items={}
    def add_item(self,name,price):
        self.items[name]=price
    def remove_item(self,name):
        if name in self.items:
            del self.items[name]
    def total_price(self):
        return sum(self.items.values())
    def display(self):
        print(self.items)

c=ShoppingCart()
c.add_item("pen",10)
c.add_item("book",50)
c.display()
print(c.total_price())
```

```python
#9. Laptop Class
#Instance variables: brand,ram,price
#Class variable: discount
#Method to apply discount
#Method to display details

class Laptop:
    discount=10
    def __init__(self,brand,ram,price):
        self.brand=brand
        self.ram=ram
        self.price=price
    def apply_discount(self):
        self.price=self.price-(self.price*Laptop.discount/100)
    def display(self):
        print(self.brand,self.ram,self.price)
l=Laptop("Dell","8GB",50000)
l.apply_discount()
l.display()
```

```python
#10. Library Management System
#Book: book_id,title,author,available
#Library: add,issue,return,display
class Book:
    def __init__(self,id,title,author):
        self.book_id=id
        self.title=title
        self.author=author
        self.available=True
class Library:
    def __init__(self):
        self.books=[]
    def add_book(self,book):
        self.books.append(book)
    def issue_book(self,id):
        for b in self.books:
            if b.book_id==id and b.available:
                b.available=False
    def return_book(self,id):
        for b in self.books:
            if b.book_id==id:
                b.available=True
    def display(self):
        for b in self.books:
            if b.available:
                print(b.title)
l=Library()
b1=Book(1,"Python","ABC")
b2=Book(2,"Java","XYZ")

l.add_book(b1)
l.add_book(b2)

l.issue_book(1)
l.display()
```

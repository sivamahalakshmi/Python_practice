```python

# 1. Bank Account Security System
# Create a BankAccount class:
# •	Private attributes: __account_number, __balance
# •	Public methods: deposit(), withdraw(), get_balance()
# •	Ensure balance cannot be accessed or modified directly.

class BankAccount:
    def __init__(self,account_number,balance):
        self.__account_number=account_number
        self.__balance=balance
    def deposit(self,amount):
        self.__balance+=amount
        print(f"amount deposited :{amount}")
    def withdraw(self,amount):
        if amount> self.__balance:
            print("insufficient balance")
        else:
            self.__balance-=amount
            print(f"amount withdrawed: {amount}")

    def get_balance(self):
        return self.__balance
a=BankAccount("iob123",5000)
a.deposit(1000)
a.withdraw(2000)
print("balance:",a.get_balance())
```

```python
# 2. Student Record System
# Create a Student class:
# •	Private attributes: __name, __marks
# •	Methods to set marks (only 0–100 allowed)
# •	Method to calculate grade
# •	Prevent invalid data assignment.
class Student:
    def __init__(self,name, marks):
        self.__name=name
        self.__marks=marks
    def setMarks(self, marks):
        if 0<= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")
    def grade(self):
        if self.__marks>=90:
            print("A")
        elif self.__marks>=80:
            print("B")
        elif self.__marks>=70:
            print("C")
        elif 0<=self.__marks<70:
            print("D")
        else:
            print("invalid data")

s=Student("maha",95)
s.grade()
```

```python
# 4. Password Protected Profile
# Create a UserProfile class:
# •	Private attribute: __password
# •	Method to change password (must verify old password)
# •	Method to validate login
# •	Do not allow direct password access.
class UserProfile:
    def __init__(self,username,password):
        self.username=username
        self.__password=password
    def login(self,password):
        if password==self.__password:
            print("login successful")
        else:
            print("invalid password")
    def change(self,old_password,new_password):
        if old_password==self.__password:
            self.__password=new_password
            print("password changed successfully")
        else:
            print("incorrect old password")
    def get_password(self):
        return self.__password
a=UserProfile("maha","maha123")
a.login("maha123")
a.change("maha123","maha456")
a.get_password()
```

```python
# 5. Product Inventory
# Create a Product class:
# •	Private attributes: __price, __stock
# •	Method to update stock (cannot be negative)
# •	Method to apply discount (cannot exceed 50%)
# •	Getter methods for price and stock.
class Product:
    def __init__(self,price,stock):
        self.__price=price
        self.__stock=stock
    def update_stock(self,stock):
        if stock>0:
            self.__stock+=stock
        else:
            print("invalid")
    def discount(self,discount):
        if 0 <=discount <=50:
            self.__price= self.__price* (1-discount/100)
        else:
            print("discount cannot exceed 50%")
    def getPrice(self):
        return self.__price
    def getStock(self):
        return self.__stock
p=Product(100,10)
p.update_stock(2)
p.discount(20)
print(p.getPrice())
print(p.getStock())
```

```python
#hierarchial inheritance - one parent, many children
class Person:
    def __init__(self,name,age,email):
        #print("i'm constructor of person")
        self.name=name
        self.age=age
        self.email=email
    def display(self):
        print(self.name,self.age,self.email)
class Student(Person):
    def __init__(self,name,age,email,roll_no):
        self.roll_no=roll_no
        super().__init__(name,age,email)
        #print("i'm constructor of student")
    def show(self):
        super().display()
        print(self.roll_no)
class Employee(Person):
    def __init__(self,name,age,email,salary):
        self.salary=salary
        super().__init__(name,age,email)
    def details(self):
        super().display()
        print(self.salary)
# s=Student("maha",20,"maha123@gmail.com",53)
# s.show()
e=Employee("maha",20,"maha123@gmail.com",100000)
e.details()
```

```python
# Single Inheritance
# 1. Person → Student
# Create a Person class with:
# •	Attributes: name, age
# •	Method: display_info()
# Create a Student class that inherits from Person:
# •	Additional attribute: roll_number
# •	Method to display student details including inherited attributes.
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print(self.name,self.age)
class Student(Person):
    def __init__(self,name,age,roll_no):
        self.roll_no=roll_no
        super().__init__(name,age)
    def show(self):
        super().display()
        print(self.roll_no)
s=Student("maha",20,53)
s.show()
```

```python
# Single Inheritance
# 2. Vehicle → Car
# Create a Vehicle class:
# •	Attributes: brand, speed
# •	Method: start()
# Create a Car class that:
# •	Inherits from Vehicle
# •	Adds attribute: fuel_type
# •	Adds method: honk()
class Vehicle:
    def __init__(self,brand,speed):
        self.brand=brand
        self.speed=speed
    def start(self):
        print("vehicle has started with speed",self.speed)
    def show(self):
        print(self.brand,self.speed)
class Car(Vehicle):
    def __init__(self,brand,speed,fuel_type):
        self.fuel_type=fuel_type
        super().__init__(brand,speed)
    def honk(self):
        super().start()
        print("car honks")
    def display(self):
        super().show()
        print(self.fuel_type)
c=Car("Rolls Royce",180,"diesel")
c.honk()
c.display()
```

```python
# Multilevel Inheritance
# 3. Animal → Dog → Puppy
# Create:
# •	Animal class with method eat()
# •	Dog class inheriting from Animal with method bark()
# •	Puppy class inheriting from Dog with method weeping()
# Demonstrate multilevel inheritance.
class Animal:
    def eat(self):
        print("animal eats")
class Dog(Animal):
    def bark(self):
        super().eat()
        print("dog barks")
class Puppy(Dog):
    def weeping(self):
        super().bark()
        print("puppy weeps")
p=Puppy()
p.weeping()
```

```python
# Hierarchical Inheritance
# #4. Shape → Rectangle & Circle
# Create a base class Shape:
# •	Attribute: color
# •	Method: display_color()
# Create two subclasses:
# •	Rectangle (area calculation)
# •	Circle (area calculation)
class Shape:
    def __init__(self,color):
        self.color=color
    def display_color(self):
        print("color:", self.color)
class Rectangle(Shape):
    def __init__(self,l,b,color):
        super().__init__(color)
        self.l=l
        self.b=b
    def area(self):
        return self.l*self.b
class Circle(Shape):
    def __init__(self,r,color):
        super().__init__(color)
        self.r=r
    def area(self):
        return 3.14*self.r**2
s=Shape("red")
r=Rectangle(2,3,"blue")
c=Circle(2,"green")
print(r.area())
print(c.area())
```

```python
# Method Overriding
# #5. Employee → Manager
# Create:
# •	Employee class with method calculate_salary()
# •	Manager class that overrides calculate_salary() to include bonus.
class Employee:
    def __init__(self,salary):
        self.salary=salary
    def calculate_salary(self):
        return self.salary
class Manager(Employee):
    def __init__(self,salary,bonus):
        super().__init__(salary)
        self.bonus=salary*(bonus/100)
    def calculate_salary(self):
        return self.salary+self.bonus
e1=Employee(20000)
o1=Manager(20000,10)
print(o1.calculate_salary())
print(e1.calculate_salary())
```

```python
# Hierarchical + Overriding
# #6. BankAccount → SavingsAccount & CurrentAccount
# Create: • BankAccount with deposit() and withdraw() • SavingsAccount with interest calculation • CurrentAccount with overdraft facility Override withdrawal behavior differently in both subclasses.
class BankAccount:
    def __init__(self,balance):
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
    def withdraw(self,amount):
        if amount<=self.balance:
            self.balance-=amount

class SavingsAccount(BankAccount):
    def __init__(self,balance,rate):
        super().__init__(balance)
        self.rate=rate
    def calculate_interest(self):
        return self.balance*(self.rate/100)
    def withdraw(self,amount):
        if amount<=self.balance:
            self.balance-=amount
        else:
            print("Insufficient balance")

class CurrentAccount(BankAccount):
    def __init__(self,balance,limit):
        super().__init__(balance)
        self.limit=limit
    def withdraw(self,amount):
        if amount<=self.balance+self.limit:
            self.balance-=amount
        else:
            print("Overdraft limit exceeded")

s=SavingsAccount(10000,5)
c=CurrentAccount(5000,2000)
s.withdraw(12000)
c.withdraw(2000)
print(s.balance)
print(c.balance)
```

```python
#Single inheritance
# 8. Appliance → WashingMachine (Using super())
# Create:
# •	Appliance class with brand and power
# •	WashingMachine inheriting from it
# •	Use super() to initialize parent attributes.
class Appliance:
    def __init__(self,brand,power):
        self.brand=brand
        self.power=power

class WashingMachine(Appliance):
    def __init__(self,brand,power,capacity):
        super().__init__(brand,power)
        self.capacity=capacity

w=WashingMachine("LG",500,7)
print(w.brand)
print(w.power)
print(w.capacity)
```

```python
# 9. Account → FixedDepositAccount (Constructor Inheritance)
# Create:
# •	Account class with account_number and balance
# •	FixedDepositAccount with additional interest_rate
# # •	Add method to calculate maturity amount.
class Account:
    def __init__(self,account_number,balance):
        self.account_number=account_number
        self.balance=balance

class FixedDepositAccount(Account):
    def __init__(self,account_number,balance,rate):
        super().__init__(account_number,balance)
        self.rate=rate
    def maturity_amount(self,time):
        return self.balance*(1+(self.rate/100)*time)

f=FixedDepositAccount(101,10000,6)
print(f.maturity_amount(2))
```

```python
# 10. Online Order System (Real-World Scenario)
# Create:
# •	Order class with order_id, customer_name
# •	OnlineOrder subclass with delivery_address
# •	InStoreOrder subclass with store_location
# •	Override a method process_order() differently
class Order:
    def __init__(self,order_id,customer_name):
        self.order_id=order_id
        self.customer_name=customer_name
    def process_order(self):
        print("Processing order")

class OnlineOrder(Order):
    def __init__(self,order_id,customer_name,address):
        super().__init__(order_id,customer_name)
        self.address=address
    def process_order(self):
        print("Delivering to",self.address)

class InStoreOrder(Order):
    def __init__(self,order_id,customer_name,location):
        super().__init__(order_id,customer_name)
        self.location=location
    def process_order(self):
        print("Pickup at",self.location)

o1=OnlineOrder(1,"maha","collector nagar")
o2=InStoreOrder(2,"susila","SS briyani")
o1.process_order()
o2.process_order()
```

```python
# UC-014: Insurance Policy Management System
# Strictly based on question paper only

class Policy:
    def __init__(self, policyNumber, holderId, type,
                 coverageAmount, premium, startDate, endDate):
        self.policyNumber = policyNumber
        self.holderId = holderId
        self.type = type
        self.coverageAmount = coverageAmount
        self.premium = premium
        self.startDate = startDate
        self.endDate = endDate
        self.status = "inactive"        # default from QP

    def activate(self):
        self.status = "active"
        print(f"Policy {self.policyNumber} activated")

    def lapse(self):
        # A2: customer misses premium → lapse after grace period
        self.status = "lapsed"
        print(f"Policy {self.policyNumber} lapsed")

    def renew(self, newEndDate):
        # main flow step 10: policy expiry → renewal
        self.endDate = newEndDate
        self.status = "active"
        print(f"Policy {self.policyNumber} renewed till {newEndDate}")

    def surrender(self):
        # A3: customer surrenders before maturity
        self.status = "surrendered"
        print(f"Policy {self.policyNumber} surrendered")

    def display(self):
        print(f"Policy: {self.policyNumber} | Holder: {self.holderId} "
              f"| Type: {self.type} | Coverage: {self.coverageAmount} "
              f"| Status: {self.status}")


# ================================================

class Claim:
    registeredIncidents = []    # for E3: duplicate claim check

    def __init__(self, claimId, policyNumber,
                 incidentDate, amount):
        self.claimId = claimId
        self.policyNumber = policyNumber
        self.incidentDate = incidentDate
        self.amount = amount
        self.status = "pending"
        self.adjusterNotes = ""

    def submit(self, policy):
        # E2: claim on lapsed policy
        if policy.status == "lapsed":
            print("Policy is lapsed. Please renew first.")
            return

        # E3: duplicate claim for same incident
        if self.incidentDate in Claim.registeredIncidents:
            print("Claim already registered for this incident.")
            return

        # E4: claim exceeds coverage → auto cap
        if self.amount > policy.coverageAmount:
            print(f"Claim exceeds coverage. "
                  f"Auto-capped at {policy.coverageAmount}")
            self.amount = policy.coverageAmount

        Claim.registeredIncidents.append(self.incidentDate)
        self.status = "submitted"
        print(f"Claim {self.claimId} submitted")

    def approve(self):
        # main flow step 9: adjuster approves
        self.status = "approved"
        print(f"Claim {self.claimId} approved")

    def reject(self, reason):
        # main flow step 9: adjuster rejects
        self.status = "rejected"
        self.adjusterNotes = reason
        print(f"Claim {self.claimId} rejected: {reason}")

    def calculateSettlement(self):
        # only if approved
        if self.status == "approved":
            print(f"Settlement for claim "
                  f"{self.claimId}: {self.amount}")
            return self.amount
        print("Claim not approved")
        return 0

    def display(self):
        print(f"Claim: {self.claimId} | "
              f"Policy: {self.policyNumber} | "
              f"Amount: {self.amount} | "
              f"Status: {self.status}")


# ================================================

class Premium:
    def __init__(self, premiumId, policyNumber,
                 dueDate, amount):
        self.premiumId = premiumId
        self.policyNumber = policyNumber
        self.dueDate = dueDate
        self.amount = amount
        self.paidDate = None
        self.status = "unpaid"

    def pay(self, paidDate, policy):
        # main flow step 6: customer pays first premium
        self.paidDate = paidDate
        self.status = "paid"
        policy.activate()
        print(f"Premium {self.premiumId} "
              f"of Rs {self.amount} paid")

    def markOverdue(self, policy):
        # A2: customer misses premium payment
        self.status = "overdue"
        policy.lapse()
        print(f"Premium {self.premiumId} overdue")

    def applyGracePeriod(self):
        # A2: grace period of 30 days from QP
        print(f"Grace period of 30 days applied "
              f"for premium {self.premiumId}")

    def display(self):
        print(f"Premium: {self.premiumId} | "
              f"Due: {self.dueDate} | "
              f"Amount: Rs {self.amount} | "
              f"Status: {self.status}")


# ================================================

class RiskEngine:
    def __init__(self, ageFactor, coverageFactor, historyFactor):
        self.ageFactor = ageFactor
        self.coverageFactor = coverageFactor
        self.historyFactor = historyFactor

    def calculatePremium(self, customer, coverage,
                          preExistingConditions=0,
                          claimFreeYears=0):
        # Section 9.3: Premium Calculation Model

        # Step 1: Base Premium
        # Coverage Amount x Base Rate (0.5%)
        basePremium = coverage * 0.005
        print(f"Base Premium      : Rs {basePremium}")

        # Step 2: Age Loading
        # 18-35: 0x | 36-50: 0.5x | 51-60: 1x | 61+: 2x
        if customer.age <= 35:
            ageLoading = basePremium * 0
        elif customer.age <= 50:
            ageLoading = basePremium * 0.5
        elif customer.age <= 60:
            ageLoading = basePremium * 1
        else:
            ageLoading = basePremium * 2
        print(f"Age Loading       : Rs {ageLoading}")

        # Step 3: Pre-existing conditions
        # +Rs 500 per declared condition per year
        preExistingLoad = preExistingConditions * 500
        print(f"Pre-existing Load : Rs {preExistingLoad}")

        # Step 4: No-claim Bonus
        # -5% for each claim-free year (max 50%)
        ncbPercent = min(claimFreeYears * 5, 50)
        ncbDiscount = (ncbPercent / 100) * basePremium
        print(f"NCB Discount      : -{ncbPercent}% "
              f"= -Rs {ncbDiscount}")

        # Step 5: Final Premium
        # Sum of above factors rounded to nearest Rs 10
        total = (basePremium + ageLoading
                 + preExistingLoad - ncbDiscount)
        finalPremium = round(total / 10) * 10
        print(f"Final Premium     : Rs {finalPremium}/yr")
        return finalPremium

    def scoreRisk(self, customer):
        # used by underwriter to assess risk
        if customer.age > 60:
            return "high"
        elif customer.age > 40:
            return "medium"
        else:
            return "low"


# ================================================

class Underwriter:
    def __init__(self, underwriterId, name):
        self.underwriterId = underwriterId
        self.name = name

    def assessRisk(self, customer, coverage,
                    preExisting=0, claimFreeYears=0):
        # main flow step 2 & 3
        risk = RiskEngine(0.005, 0.005, 0.005)
        riskScore = risk.scoreRisk(customer)
        premium = risk.calculatePremium(
            customer, coverage,
            preExisting, claimFreeYears
        )
        return premium, riskScore

    def approve(self, policy):
        # main flow step 5
        policy.activate()
        print(f"Underwriter {self.name} "
              f"approved policy {policy.policyNumber}")

    def reject(self, policy, reason):
        # E1: application rejected — high risk profile
        policy.status = "rejected"
        print(f"Application rejected: "
              f"high risk profile — {reason}")


# ================================================

class Customer:
    def __init__(self, customerId, name, age, kycStatus):
        self.customerId = customerId
        self.name = name
        self.age = age
        self.kycStatus = kycStatus
        self.policies = []

    def applyPolicy(self, policy, underwriter,
                     coverage, preExisting=0,
                     claimFreeYears=0):
        # E5: KYC documents missing
        if self.kycStatus != "verified":
            print(f"Application on hold: "
                  f"KYC pending for {self.name}")
            return

        # main flow step 2,3,4
        premium, riskScore = underwriter.assessRisk(
            self, coverage, preExisting, claimFreeYears
        )

        # E1: high risk → reject
        if riskScore == "high":
            underwriter.reject(policy, "high risk profile")
            return

        # main flow step 5: approved
        policy.premium = premium
        underwriter.approve(policy)
        self.policies.append(policy)
        print(f"{self.name} enrolled in "
              f"policy {policy.policyNumber}")

    def raiseClaim(self, claim):
        # main flow step 7
        for policy in self.policies:
            if policy.policyNumber == claim.policyNumber:
                claim.submit(policy)
                return
        print("No matching policy found")

    def renewPolicy(self, policyNumber, newEndDate):
        # main flow step 10
        for policy in self.policies:
            if policy.policyNumber == policyNumber:
                policy.renew(newEndDate)
                return
        print("Policy not found")

    def display(self):
        print(f"Customer: {self.customerId} | "
              f"Name: {self.name} | "
              f"Age: {self.age} | "
              f"KYC: {self.kycStatus}")
        print(f"Policies: "
              f"{[p.policyNumber for p in self.policies]}")


# ================================================
# TEST — following exact main flow from QP
# ================================================

print("=" * 50)
print("UC-014: INSURANCE POLICY MANAGEMENT SYSTEM")
print("=" * 50)

# Underwriter
uw1 = Underwriter("U001", "Ravi")

# --- Main Flow ---
print("\n--- Step 1: Customer applies ---")
c1 = Customer("C001", "Maha", 42, "verified")
c1.display()

p1 = Policy("P001", "C001", "Health",
            500000, 0,
            "2024-01-01", "2025-01-01")

# Steps 2,3,4,5: assess risk, calculate premium, approve
print("\n--- Steps 2-5: Risk & Approval ---")
c1.applyPolicy(p1, uw1,
               coverage=500000,
               preExisting=2,
               claimFreeYears=3)
c1.display()

# Step 6: pay premium
print("\n--- Step 6: Premium Payment ---")
pr1 = Premium("PR001", "P001", "2024-02-01", p1.premium)
pr1.pay("2024-02-01", p1)
pr1.display()

# Step 7: raise claim
print("\n--- Step 7: Raise Claim ---")
cl1 = Claim("CL001", "P001", "2024-06-15", 50000)
c1.raiseClaim(cl1)

# Steps 8,9: adjuster reviews
print("\n--- Steps 8-9: Claim Processing ---")
cl1.approve()
cl1.calculateSettlement()
cl1.display()

# Step 10: renewal
print("\n--- Step 10: Policy Renewal ---")
c1.renewPolicy("P001", "2026-01-01")
p1.display()

# --- Exception Flows ---
print("\n--- E5: KYC Missing ---")
c2 = Customer("C002", "Kumar", 35, "pending")
p2 = Policy("P002", "C002", "Vehicle",
            200000, 0,
            "2024-01-01", "2025-01-01")
c2.applyPolicy(p2, uw1, 200000)

print("\n--- E2: Claim on Lapsed Policy ---")
p1.lapse()
cl2 = Claim("CL002", "P001", "2024-08-01", 20000)
cl2.submit(p1)

print("\n--- E3: Duplicate Claim ---")
cl3 = Claim("CL003", "P001", "2024-06-15", 10000)
cl3.submit(p1)

print("\n--- E4: Claim Exceeds Coverage ---")
p1.activate()
cl4 = Claim("CL004", "P001", "2024-10-01", 9999999)
cl4.submit(p1)

print("\n--- A2: Premium Overdue → Grace Period ---")
pr2 = Premium("PR002", "P001", "2024-03-01", p1.premium)
pr2.applyGracePeriod()
pr2.markOverdue(p1)

print("\n--- A3: Policy Surrender ---")
p1.activate()
p1.surrender()
```

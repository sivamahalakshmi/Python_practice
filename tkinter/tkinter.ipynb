```python
import tkinter as tk
from tkinter import messagebox, filedialog

# Function to display messages
def show_message(message):
    messagebox.showinfo("Menu Action", message)

# Function to handle "New" menu action
def new_file():
    text_area.delete(1.0, tk.END)  # Clear the text area

# Function to handle "Open" menu action
def open_file():
    file_path = filedialog.askopenfilename(
        title="Open File",
        filetypes=[("Text Files", "*.txt"), ("All Files", "*.*")]
    )
    if file_path:
        try:
            with open(file_path, "r") as file:
                content = file.read()
                text_area.delete(1.0, tk.END)  # Clear the text area
                text_area.insert(tk.END, content)  # Load content into the text area
        except Exception as e:
            messagebox.showerror("Error", f"Failed to open file: {e}")

# Function to handle "Save" menu action
def save_file():
    file_path = filedialog.asksaveasfilename(
        title="Save File",
        defaultextension=".txt",
        filetypes=[("Text Files", "*.txt"), ("All Files", "*.*")]
    )
    if file_path:
        try:
            with open(file_path, "w") as file:
                content = text_area.get(1.0, tk.END)  # Get all text from the text area
                file.write(content.strip())  # Save content to the file
        except Exception as e:
            messagebox.showerror("Error", f"Failed to save file: {e}")

# Create the main application window
root = tk.Tk()
root.title("Menu Example with Text Area and File Chooser")
root.geometry("600x400")  # Set window size

# Create the menu bar
menu_bar = tk.Menu(root)

# Add a "File" menu
file_menu = tk.Menu(menu_bar, tearoff=0)
file_menu.add_command(label="New", command=new_file)
file_menu.add_command(label="Open", command=open_file)
file_menu.add_command(label="Save", command=save_file)
file_menu.add_separator()
file_menu.add_command(label="Exit", command=root.quit)
menu_bar.add_cascade(label="File", menu=file_menu)

# Add an "Edit" menu
edit_menu = tk.Menu(menu_bar, tearoff=0)
edit_menu.add_command(label="Cut", command=lambda: text_area.event_generate("<<Cut>>"))
edit_menu.add_command(label="Copy", command=lambda: text_area.event_generate("<<Copy>>"))
edit_menu.add_command(label="Paste", command=lambda: text_area.event_generate("<<Paste>>"))
menu_bar.add_cascade(label="Edit", menu=edit_menu)

# Add a "Help" menu
help_menu = tk.Menu(menu_bar, tearoff=0)
help_menu.add_command(label="About", command=lambda: show_message("This is a demo application with file chooser!"))
menu_bar.add_cascade(label="Help", menu=help_menu)

# Attach the menu bar to the root window
root.config(menu=menu_bar)

# Add a text area
text_area = tk.Text(root, wrap=tk.CHAR, font=("Arial", 12))
text_area.pack(expand=True, fill=tk.BOTH)  # Make the text area expand and fill the window

# Run the Tkinter event loop
root.mainloop()
```

```python
import tkinter as tk

def show_selected():
    # curselection() returns a tuple of indices
    selected_indices = lb.curselection()
    for i in selected_indices:
        print(f"Selected: {lb.get(i)}")

root = tk.Tk()
lb = tk.Listbox(root, selectmode='multiple')
lb.pack()

# Adding items
items = ["Python", "Java", "C++", "JavaScript"]
for item in items:
    lb.insert(tk.END, item)

btn = tk.Button(root, text="Submit", command=show_selected)
btn.pack()

root.mainloop()
```

```python
import mysql.connector
import tkinter as tk
from tkinter import messagebox
from tkinter import ttk

# Establish the connection
conn = mysql.connector.connect(
    host="localhost",  # Your MySQL host (use 'localhost' or an IP address)
    user="root",  # Your MySQL username
    password="root",  # Your MySQL password
    database="ssi"  # Your database name
)

# Create a cursor object to interact with the database
cursor = conn.cursor()

# --------------------------
# SELECT - Fetch records
# --------------------------
def select_data():
    cursor.execute("SELECT * FROM student")
    rows = cursor.fetchall()
    for row in rows:
        tree.insert("", "end", values=row)

# --------------------------
# INSERT - Add new record
# --------------------------
def insert_data():
    try:
        id = int(entry_id.get())
        name = entry_name.get()
        course = entry_course.get()

        sql_insert = """
        INSERT INTO student (id, name, course)
        VALUES (%s, %s, %s)
        """
        data = (id, name, course)
        cursor.execute(sql_insert, data)
        conn.commit()  # Commit the transaction
        messagebox.showinfo("Success", "Record inserted successfully.")
        clear_entries()  # Clear text fields after insertion

    except Exception as e:
        messagebox.showerror("Error", f"Error inserting data: {e}")

# --------------------------
# DELETE - Remove a record
# --------------------------
def delete_data():
    try:
        id = int(entry_delete_id.get())
        sql_delete = """
        DELETE FROM student
        WHERE id = %s
        """
        data = (id,)
        cursor.execute(sql_delete, data)
        conn.commit()  # Commit the transaction
        messagebox.showinfo("Success", "Record deleted successfully.")
        entry_delete_id.delete(0, tk.END)  # Clear the input field
    except Exception as e:
        messagebox.showerror("Error", f"Error deleting data: {e}")

# --------------------------
# Clear input fields
# --------------------------
def clear_entries():
    entry_id.delete(0, tk.END)
    entry_name.delete(0, tk.END)
    entry_course.delete(0, tk.END)

# --------------------------
# Setup Tkinter UI
# --------------------------

root = tk.Tk()
root.title("Student Records Management")

# --------------------------
# Treeview to display records
# --------------------------
tree = ttk.Treeview(root, columns=("ID", "Name", "Course"), show="headings")
tree.heading("ID", text="ID")
tree.heading("Name", text="Name")
tree.heading("Course", text="Course")
tree.pack(pady=20)

# Button to display records
btn_fetch = tk.Button(root, text="Fetch Records", command=select_data)
btn_fetch.pack(pady=10)

# --------------------------
# Insert Data Section
# --------------------------
frame_insert = tk.LabelFrame(root, text="Insert New Record", padx=10, pady=10)
frame_insert.pack(padx=10, pady=10, fill="both", expand="yes")

tk.Label(frame_insert, text="ID:").grid(row=0, column=0, padx=5, pady=5)
tk.Label(frame_insert, text="Name:").grid(row=1, column=0, padx=5, pady=5)
tk.Label(frame_insert, text="Course:").grid(row=2, column=0, padx=5, pady=5)

entry_id = tk.Entry(frame_insert)
entry_name = tk.Entry(frame_insert)
entry_course = tk.Entry(frame_insert)

entry_id.grid(row=0, column=1, padx=5, pady=5)
entry_name.grid(row=1, column=1, padx=5, pady=5)
entry_course.grid(row=2, column=1, padx=5, pady=5)

btn_insert = tk.Button(frame_insert, text="Insert Record", command=insert_data)
btn_insert.grid(row=3, columnspan=2, pady=10)

# --------------------------
# Delete Data Section
# --------------------------
frame_delete = tk.LabelFrame(root, text="Delete Record by ID", padx=10, pady=10)
frame_delete.pack(padx=10, pady=10, fill="both", expand="yes")

tk.Label(frame_delete, text="ID:").grid(row=0, column=0, padx=5, pady=5)

entry_delete_id = tk.Entry(frame_delete)
entry_delete_id.grid(row=0, column=1, padx=5, pady=5)

btn_delete = tk.Button(frame_delete, text="Delete Record", command=delete_data)
btn_delete.grid(row=1, columnspan=2, pady=10)

# Start Tkinter event loop
root.mainloop()

# Close the cursor and connection after the GUI is closed
cursor.close()
conn.close()
```

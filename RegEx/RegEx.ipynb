```python
# 1. Find Digits
# Write a Python regex program to find all digits in the string.
# Example:
# text = "My marks are 78 and 95"
# Expected Output:
# ['78', '95']
import re
s=input("enter string:")
x=re.findall("[0-9]+",s)
print(x)
```

```python
# 2. Find Words Starting with a
# Write a regex to find all words starting with letter a.
# Example:
# text = "apple ball ant cat"
import re
s=input("enter string:")
x=re.findall("\\ba\\w*",s)
print(x)
```

```python
# 3. Replace Spaces with Hyphen
# Use regex to replace all spaces with -.
# Example:
# text = "Python is easy"
# Expected Output:
# Python-is-easy
import re
s=input("enter string:")
x=re.sub("\\s","-",s)
print(x)
```

```python
# 4. Count Number of Vowels
# Write a regex program to count vowels in a string.
# Example:
# text = "education"
import re
s=input("enter string:")
x=re.findall("[aeiouAEIOU]",s)
print("vowels:",x)
print("count:",len(x))
```

```python
# 5. Find Email IDs
# Write a regex to extract email addresses from text.
# Example:
# text = "Contact us at abc@gmail.com and test@yahoo.com"
import re
email=input("enter string:")
x=re.findall("[a-zA-Z]\\w*@[a-zA-Z]+\\.[a-zA-Z]+",email)
print(x)
```

```python
# 6. Check Mobile Number
# Write a regex to check whether a mobile number contains exactly 10 digits.
# Example:
# number = "9876543210"
import re
mobile=input("enter mobile number:")
x=re.findall('[0-9]{10}',mobile)
print(x)
```

```python
# 7. Find Words Ending with ing
# Write a regex to find all words ending with ing.
# Example:
# text = "I am playing and singing"
import re
string=input("enter string:")
x = re.findall(r"\b\w+ing\b", string)
print(x)
```

```python
# 8. Remove All Digits from a String
# Use regex to remove numbers from a string.
# Example:
# text = "abc123xyz45"
# Expected Output:
# abcxyz
import re
s = input("Enter string: ")
x = re.sub("[0-9]", "", s)
print(x)
```

```python
# 9. Split Sentence into Words
# Use re.split() to split a sentence using spaces.
# Example:
# text = "Python Regular Expressions"
import re
s = input("Enter sentence: ")
x = re.split(" ", s)
print(x)
```

```python
# 10. Find Repeated Words
# Write a regex to find repeated words.
# Example:
# text = "hello hello world world"
# Expected Output:
# ['hello', 'world']
import re
s = input("Enter string: ")
x = re.findall(r"\b(\w+)\s+\1\b", s)
print(x)
```

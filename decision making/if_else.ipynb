```python
#1.Even or Odd Number
n=int(input("Enter a number:"))
if n%2==0:
    print(f"{n} is a even number")
else:
    print(f"{n} is a odd number")
```

```python
#2.Positive, Negative, or Zero
n=int(input("Enter a number:"))
if n==0:
    print(f"The number is zero")
elif n>0:
    print("The number is positive")
else:
    print("The number is negative")
```

```python
#3.Largest of Two Numbers
a=int(input("Enter a:"))
b=int(input("Enter b:"))
if a>b:
    print(f"{a} is larger")
else:
    print(f"{b} is larger")
```

```python
#4.Age Group Classifier
age=int(input("Enter age:"))
if age <=12:
    print("Child")
elif age<=19:
    print("Teenager")
elif age<64:
    print("Adult")
else:
    print("Senior")
```

```python
#5.Grade Calculator
marks=int(input("Enter marks"))
if marks>=90:
    print("A")
elif marks>=80:
    print("B")
elif marks>=70:
    print("C")
elif marks>=60:
    print("D")
else:
    print("F")
```

```python
#6.Number in Range
num=int(input("Enter a number:"))
if num<=50:
    print("Number in range 0-50")
elif num<100:
    print("Number in range 51-100")
elif num<200:
    print("Number in range 101-200")
else:
    print("Number is greater than 200")
```

```python
#7.Leap Year Checker
year=int(input("enter year:"))
if year%400==0 or (year%4==0 and year%100!=0):
    print("leap year")
else:
    print("not a leap year")
```

```python
#8.Simple Calculator
a=int(input("Enter first number:"))
b=int(input("Enter second number"))
c=input("Enter operator (+,-,*,/):")
if c=="+":
    print(a+b)
elif c=="-":
    print(a-b)
elif c=="*":
    print(a*b)
elif c=="/":
    print(a/b)
```

```python
#9.Vowel or Consonant
char=input("enter a character")
vowel="aeiouAEIOU"
if char in vowel:
    print(f"{char} is a vowel")
else:
    print(f"{char} is a consonant")
```

```python
#11.Odd or Even Range
num=int(input("Enter a number"))
if num%2==0:
    if num<=100:
        print("The number is even and between 1 and 100")
else:
    if num<=100:
        print("The number is odd and between 1-100")
```

```python
#12.Character type checker
ch=input("Enter a character")
if ch.isupper():
    print(ch, "is an uppercase letter.")
elif ch.islower():
    print(ch, "is a lowercase letter.")
elif ch.isdigit():
    print(ch, "is a digit.")
else:
    print(ch, "is a special character.")
```

```python
#13.Triangle Type Checker
a,b,c=map(int,input("Enter sides:").split())
if a==b==c:
    print("Equilateral triangle")
elif a==b or b==c or c==a:
    print("Isosceles triangle")
else:
    print("Scalene triangle")
```

```python
#14.Divisible by 5 and 11
num=int(input())
if num%5==0 and num%11==0:
    print(f"{num} is  divisible by both 5 and 11")
else:
    print(f"{num} is not divisible by 5 and 11")
```

```python
#15.Biggest of 3 numbers
a = int(input("Enter a:"))
b = int(input("Enter b:"))
c = int(input("Enter c:"))
if a >= b and a >= c:
    print(a, "is big")
elif b >= a and b >= c:
    print(b, "is big")
else:
    print(c, "is big")
```

```python
#19.Traffic Light Signal
clr=input("Enter color (red/yellow/green")
if clr=="red":
    print("Stop")
elif clr=="yellow":
    print("wait")
else:
    print("go")
```

```python
#20. Electricity bill calculator
n=int(input("Enter the units consumed: "))
if n<=100:
    bill=n*5
elif n <= 200:
    bill = (100*5)+(n-100)*7
else:
    bill = (100*5)+(100*7)+(n-200)*10
print("Bill amount:", bill)
```

```python
#21. Ten's place odd or even
n = int(input("Enter a number: "))
tens = (n // 10) % 10
if tens%2==0:
    print(tens, "is even")
else:
    print(tens, "is odd")
```

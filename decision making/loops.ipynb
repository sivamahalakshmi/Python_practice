```python
#1.Print 1 to N
n=int(input())
for i in range(1,n+1):
    print(i)
```

```python
#2.Print all even numbers up to N
n=int(input())
for i in range(2,n+1,2):
    print(i)
```

```python
#3.Sum of digits
n=int(input())
sum=0 #123
while n>0:
     sum+=n%10
     n=n//10
print(sum)
```

```python
#4.Reverse a number
n=int(input()) #1234
rev=0
while n>0:
    r=n%10
    rev=rev*10+r
    n=n//10
print(rev)
```

```python
#5.Count number of digits
n=int(input())
count=0
while n>0:
    c=n%10
    if c!=0:
        count+=1
    n//=10
print(count)
```

```python
#6.Factorial of N
n=int(input())
fact=1
for i in range(1,n+1):
    fact*=i
print(fact)
```

```python
#7.Palindrome number
n=int(input()) #121
temp=n
pal=0
while n>0:
    r=n%10
    pal=pal*10+r
    n=n//10
if pal==temp:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#8.Sum of even digits
n=int(input())
sum=0 #123
while n>0:
     dig=n%10
     if dig%2==0:
         sum+=dig
     n=n//10
print(sum)
```

```python
#9.Largest digit in a number
n=int(input()) #1234
big=0
while n>0:
    d=n%10
    if d>big:
        big=d
    n//=10
print(big)
```

```python
#10.Print tables of a number
n=int(input())
for i in range(1,11):
    t=i*n
    print(t)
```

```python
#11.Print digits in reverse order
n=int(input())
pal=0
while n>0:
    r=n%10
    pal=pal*10+r
    n=n//10
print(pal)
```

```python
#12.count even and odd numbers
n=int(input())
even=0
odd=0
while n>0:
    r=n%10
    if r%2==0:
        even+=1
    else:
        odd+=1
    n//=10
print(f"even:{even} odd:{odd}")
```

```python
#13.power using while
b,e=map(int,input().split())
res=1
while e>0:
    res*=b
    e-=1
print(res)
```

```python
#14.Armstrong number using while
n=int(input())
temp=n
s=0
while n>0:
    d=n%10
    s+=d**3
    n//=10
print("Armstrong Number" if s == temp else "Not Armstrong")
```

```python
#15.Fibonacci
n=int(input())
a, b = 0, 1
for i in range(n):
    print(a,end=" ")
    a,b=b,a+b
```

```python
#16.GCD
a,b = map(int, input().split())
while b!=0:
    a,b=b,a%b
print(a)
```

```python
#17.prime
n = int(input())
count=0
for i in range(2,n):
    if n%i==0:
        count+=1
print("not prime" if count>0 else "prime")
```

```python
#18.LCM
a,b=map(int,input().split())
x,y = a,b
while y:
    x,y=y,x%y
gcd=x
print((a*b)//gcd)
```

```python
#19. product of digits
n=int(input())
p=1
while n>0:
    p*=n%10
    n//=10
print(p)
```

```python
#20.binary
n=int(input())
b=""
while n>0:
    b=str(n%2)+b
    n//=2
print(b)
```

```python
#21.square root
n=int(input())
i=1
while i*i<=n:
    i+=1
print(i-1)
```

```python
#22.Harshad
n=int(input())
temp=n
s=0
while n>0:
    s+=n%10
    n//= 10
print("Harshad Number" if temp%s==0 else "Not Harshad")
```

```python
#23.count zeros
n=int(input())
c=0
while n>0:
    if n%10==0:
        c+=1
    n//=10
print(c)
```

```python
#24.decimal to octal
n=int(input())
o=""
while n>0:
    o=str(n%8)+o
    n//=8
print(o)
```

```python
#25.n stars
n=int(input())
for i in range(n):
    print("*",end="")
```

```python
#26. right triangle
n = int(input())
for i in range(1, n+1):
    print("*"*i)
```

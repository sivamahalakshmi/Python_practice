```python
#user defined iterator
class Fibonacci:
    def __init__(self,n):
        self.n=n
        self.a=0 #fibonacci first value
        self.b=1  #fibonacci second value
        self.count =0 #counts the number of fibonnaci numbers used so for
    #step 1: create the __iter__(self) method


    def __iter__(self):
      return self #mandatory because as per python documentation __iter__() function should always return self
    #step 2: create the __next__(self) method
    def __next__(self):
      if self.count  <= self.n :
          fib=self.a
          self.a,self.b=self.b,self.a+self.b
          self.count=self.count+1
          return fib  #it returns a single value
      else:
          raise StopIteration
# 0 1 1 2
n=int(input("Enter the limit"))
f=Fibonacci(n) # the __iter__method is called and an object is returned to f
for var in f:
    print(var)
```

```python
# Write a user-defined iterator that generates numbers from 1 to n. Use the _iter() and __next_() methods.
class Numbers:
    def __init__(self,n):
        self.n=n
        self.num=1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num<=self.n:
            value=self.num
            self.num=self.num+1
            return value
        else:
            raise StopIteration

n=int(input("Enter the limit"))
f=Numbers(n)

for var in f:
    print(var)
```

```python
# Create a custom iterator class EvenNumbers that iterates through even numbers between 1 and 20.
# Demonstrate iteration using a for loop.
class EvenNumbers:
    def __init__(self):
        self.num=2

    def __iter__(self):
        return self

    def __next__(self):
        if self.num<=20:
            even=self.num
            self.num=self.num+2
            return even
        else:
            raise StopIteration

e=EvenNumbers()
for var in e:
    print(var)
```

```python
# .Write a Python program to create a user-defined iterator for a string that returns one character at a time.
# Handle the StopIteration exception properly.
class StringIterator:
    def __init__(self,text):
        self.text=text
        self.index=0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index<len(self.text):
            ch=self.text[self.index]
            self.index=self.index+1
            return ch #it returns a single character
        else:
            raise StopIteration

s=input("Enter a string:")
obj=StringIterator(s)
try:
    while True:
        print(next(obj))
except StopIteration:
    print("Iteration completed")
```

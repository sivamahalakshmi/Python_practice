```python
# 1.Write a Python function calculate_sum(*args) that accepts a dynamic number of integer arguments and returns their sum.
# Test the function with at least 5 different inputs.
def calculate_sum(*args):
    s=0
    for i in args:
        s+=i
    return s
calculate_sum(1,2,3,4)
```

```python
# 2.Write a function student_details(**kwargs) that accepts a dynamic number of keyword arguments and prints student information in the format:
# Key : Value
# Demonstrate the function with at least 4 keyword arguments.
def student_details(**kwargs):
    for key,value in kwargs.items():
        print(key, ":", value)
student_details(name="Maha",age=20,collge="rec")
```

```python
# 3.Create a function display_info(name, *scores, **details) that accepts:
# •	A fixed argument name
# •	A variable number of positional arguments representing marks
# •	A variable number of keyword arguments representing additional details
# Display all values clearly.
def display_info(name,*marks,**details):
    print("name:",name)
    print("\nScores:")
    for mark in marks:
        print(mark)
    print("\nDetails")
    for key,value in details.items():
        print(key,":",value)
display_info("maha",99,98,97,year="3rd",college="rec")
```

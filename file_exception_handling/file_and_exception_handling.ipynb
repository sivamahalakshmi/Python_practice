```python
# Q1. Write a Python program to read a text file data.txt and count the number of:
# Lines
# Words
# Characters
# Print the results in a neat format. [Hint : f.readlines() helps count lines.split() helps count words.len(text) gives character count.]
try:
    f=open("data.txt","w")
    f.write("hello world\npython is easy")
    f.close()

    f=open("data.txt","r")
    text=f.read()

    lines=len(text.split("\n"))
    words=len(text.split())
    chars=len(text)

    print("Lines:",lines)
    print("Words:",words)
    print("Characters:",chars)

    f.close()
except:
    print("Error")
```

```python
#Q3. Write a Python script to copy the contents of one file (source.txt) into another file (destination.txt) .
try:
    f1=open("source.txt","w")
    f1.write("hello world\npython")
    f1.close()
    f1=open("source.txt","r")
    data=f1.read()
    f2=open("destination.txt","w")
    f2.write(data)
    f1.close()
    f2.close()
    print("Copied successfully")
except:
    print("Error")
```

```python
#Q4. A file marks.txt contains student names and their marks in the following format:
# Alice 85
# Bob 67
# Charlie 92
# David 74
# Write a Python program to read the file and print:
# The student with the highest marks
# The student with the lowest marks [ HINT : Use split() to separate name and marks.
# Convert marks to int.]
try:
    f=open("marks.txt","w")
    f.write("Alice 85\nBob 67\nCharlie 92\nDavid 74")
    f.close()
    f=open("marks.txt","r")
    maxm=0
    minm=100
    maxname=""
    minname=""

    for i in f:
        name,marks=i.split()
        marks=int(marks)
        if marks>maxm:
            maxm=marks
            maxname=name
        if marks<minm:
            minm=marks
            minname=name

    print("Highest:",maxname,maxm)
    print("Lowest:",minname,minm)
    f.close()
except:
    print("Error")
```

```python
# Q5. Write a Python program to read a file story.txt and count how many times each word occurs.
# Ignore case sensitivity (The and the should be considered the same).
# Display the word frequencies in sorted order (alphabetically).[HINT : Read file content and convert to lowercase (.lower()).Use split() to break into words.Use a dictionary to count occurrences (word_count[word] += 1).
# Sort dictionary keys using sorted().
try:
    f=open("story.txt","w")
    f.write("The cat and the dog The cat")
    f.close()
    f=open("story.txt","r")
    text=f.read().lower()
    words=text.split()
    d={}
    for i in words:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    for i in sorted(d):
        print(i,d[i])
    f.close()
except:
    print("Error")
```

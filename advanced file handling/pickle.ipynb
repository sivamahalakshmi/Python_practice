```python
#serialisation
import pickle
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
s=Student("abinaya",12)
f=open("student1.pkl","wb")
pickle.dump(s,f)
f.close()
```

```python
#deserialisation
import pickle
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
f=open("student1.pkl","rb")
t=pickle.load(f)
print(t.name)
print(t.age)
f.close()
```

```python
# 1. Do pickle and unpickle for employee obj with attributes id, name, salary, department
import pickle
class Employee:
    def __init__(self,id,name,salary,department):
        self.id=id
        self.name=name
        self.salary=salary
        self.department=department
#serialisation
e=Employee("emp123","maha",70000,"IT")
f=open("employee1.pkl","wb")
pickle.dump(e,f)
f.close()

#deserialisation
f=open("employee1.pkl","rb")
n_f=pickle.load(f)
print(n_f.name)
print(n_f.salary)
f.close()
```

```python
# Write a Python program to store a list of numbers into a file using pickle and then read it back using unpickle.
# Display the original and retrieved data.
import pickle
numbers=[10,20,30,40,50]

f=open("numbers.pkl","wb")
pickle.dump(numbers,f)
f.close()

f=open("numbers.pkl","rb")
retrieved_data=pickle.load(f)
f.close()

print("Original Data:",numbers)
print("Retrieved Data:",retrieved_data)
```

```python
# # Create a Python program that:
# 1.	Stores a dictionary containing employee details into a file using pickle
# 2.	Retrieves and displays the dictionary using unpickle
import pickle
employee={"id":101,"name":"Abinaya","department":"IT","salary":45000}

f=open("employee.pkl","wb")
pickle.dump(employee,f)
f.close()

f=open("employee.pkl","rb")
data=pickle.load(f)
f.close()

print("Original Dictionary:",employee)
print("Retrieved Dictionary:",data)
```

```python
#Write a program to pickle multiple objects (list, tuple, dictionary) into a file and unpickle them one by one.
# Program to pickle multiple objects and unpickle them one by one
import pickle
mylist=[10,20,30]
mytuple=("Python","Java","C")
mydict={"name":"Abinaya","age":20}

f=open("data.pkl","wb")

pickle.dump(mylist,f)
pickle.dump(mytuple,f)
pickle.dump(mydict,f)
f.close()

#unpickling
f=open("data.pkl","rb")

list_data=pickle.load(f)
tuple_data=pickle.load(f)
dict_data=pickle.load(f)

f.close()
print("List:",list_data)
print("Tuple:",tuple_data)
print("Dictionary:",dict_data)
```

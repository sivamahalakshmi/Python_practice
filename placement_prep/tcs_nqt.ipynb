```
# 1.  Check Even or Odd
n=int(input())
if n%2==0:
    print("even")
else:
    print("odd")
```

```
## 2.  Check Prime Number
n=int(input())
c=0
for i in range(1,n+1):
    if n%i==0:
        c+=1
print("Prime" if c==2 else "Not Prime")

```

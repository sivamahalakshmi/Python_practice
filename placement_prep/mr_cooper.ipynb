```python
# Find the length of the longest word in a sentence.
def longest_Word(s):
    long=0
    s1=s.strip().split()
    for i in s1:
        check=len(i)
        if check>long:
            long=check
    return long
print(longest_Word("Find the length of the longest word in a sentence."))
```

```python
# Find the number of vowels in a given string.
def no_of_vowels(s):
    count=0
    vowels="aeiouAEIOU"
    for i in s:
        if i in vowels:
            count+=1
    return count
print(no_of_vowels("maha"))
```

```python
# Keep adding the digits of a number until it becomes a single digit. Return how many steps it took.\
def single_digit(s):
    s=int(s)
    count=0
    while s>=10:
        tot=0
        while s>0:
            a=s%10
            tot+=a
            s=s//10
        s=tot
        count+=1
    return tot,count
print(single_digit("9999"))

```

```python
# Keep multiplying the digits of a number until it becomes a single digit.
def single_digit_multiply(s):
    s=int(s)
    count=0
    while s>=10:
        tot=1
        while s>0:
            a=s%10
            if a==0:
                s//=10
                continue
            tot*=a
            s=s//10
        s=tot
        count+=1
    return s,count
print(single_digit_multiply("180"))
```

```python
# 4. Employee → Manager (Method Overriding)
# Create:
# - Employee class with method calculate_salary()
# - Manager class that overrides calculate_salary() to include bonus.
# Demonstrate method overriding.

class Employee:
    def __init__(self,salary):
        self.salary=salary
    def calculate_salary(self):
        return self.salary
class Manager(Employee):
    def __init__(self,salary,bonus):
        super().__init__(salary)
        self.bonus=salary*(bonus/100)
    def calculate_salary(self):
        return self.salary+self.bonus
e1=Employee(20000)
o1=Manager(20000,10)
print(o1.calculate_salary())
print(e1.calculate_salary())
```

```python
# **ENCAPSULATION**
#
# Bank Account Security System
# Create a BankAccount class:
# - Private attributes: __account_number, __balance
# - Public methods: deposit(), withdraw(), get_balance()
# - Ensure balance cannot be accessed or modified directly.
class BankAccount:
    def __init__(self,account_number,balance):
        self.__account_number=account_number
        self.__balance=balance
    def deposit(self,amount):
        self.__balance+=amount
        print(f"amount deposited: {amount}")
    def withdraw(self,amount):
        if amount>self.__balance:
            print("insufficient balance")
        else:
            self.__balance-=amount
            print(f"amount withdrawn: {amount}")
    def get_balance(self):
        return self.__balance
a=BankAccount("acc1024",10000)
a.deposit(5000)
a.deposit(3000)
print("Current balance:",a.get_balance())
```

```python
#1.Library management system
# Build a library system where books can be
# added, borrowed, and returned. Show all
# available books at any time.
class Book:
    def __init__(self,title,author,copies):
        self.title=title
        self.author=author
        self.copies=copies
class Library:
    def __init__(self):
        self.books=[]
    def add_book(self,book):
        self.books.append(book)
        print(f"{book.title} added")
    def borrow_book(self,title):
        for book in self.books:
            if book.title == title:
                book.copies-=1
                print(f"{book.title} borrowed")
                return
        print("book not found")
    def return_book(self,title):
        for book in self.books:
           if book.title == title:
            book.copies+=1
            print(f"{book.title} returned")
            return
        print("book not found")
    def show_books(self):
        for book in self.books:
            print(f"{book.title} | {book.author} | {book.copies}")
lib=Library()
while True:
    print("\n--- Library Menu ---")
    print("1. Add Book")
    print("2. Borrow Book")
    print("3. Return Book")
    print("4. Show Books")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        title = input("Enter title: ")
        author = input("Enter author: ")
        copies = int(input("Enter copies: "))
        book = Book(title, author, copies)
        lib.add_book(book)
    elif choice == "2":
        title = input("Enter title to borrow: ")
        lib.borrow_book(title)
    elif choice == "3":
        title = input("Enter title to return: ")
        lib.return_book(title)
    elif choice == "4":
        lib.show_books()
    elif choice == "5":
        print("Goodbye!")
        break
    else:
        print("Invalid choice")

# b1=Book("Python","John",3)
# b2=Book("OOP Basics", "Smith", 2)
# lib=Library()
# lib.add_book(b1)
# lib.add_book(b2)
# lib.show_books()
# lib.borrow_book("Python")
# lib.show_books()
```

```python
#Classes   → ParkingLot
# Attributes→ total_slots, available_slots
# Methods   → park(), remove(), show_slots()
class ParkingLot:
    def __init__(self,total_slots):
        self.total_slots=total_slots
        self.available_slots=total_slots
    def park(self,count):
        if self.available_slots>=count:
            self.available_slots-=count
            print(f"{count} vehicle(s) parked")
        else:
            print("parking is full")
    def remove(self,count):
        if self.available_slots < self.total_slots:
            self.available_slots+=count
            print(f"{count} vehicle(s) removed")
        else:
            print("parking slot empty")
    def show_slots(self):
        print(f"total slots:{self.total_slots} | available_slots:{self.available_slots}")
p1=ParkingLot(10)
p1.show_slots()
p1.park(3)
p1.show_slots()
p1.remove(1)
p1.show_slots()
p1.park(20)

```

```python
#5.find no. of even or odd divisors of a given integer
n=int(input())
even=0
odd=0
for i in range(1,n+1):
    if n%i==0:
        if n%2==0:
            even+=1
        else:
            odd+=1
print("even divisors:",even)
print("odd divisors:",odd)
```

```python
#17.factorial of a positive integer
# n=int(input())
# fact=1
# for i in range(1,n+1):
#     fact*=i
# print(fact)

#using recursion
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n * fact(n-1)
n=int(input())
print(fact(n))
```

```python
#18.fibonacci using recursion
# a=0
# b=1
# while a<50:
#     print(a,end=" ")
#     a,b=b,a+b

def fib(n):
    if n==0:
        return 0
    if n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)
n=int(input())
print(fib(n))
```
